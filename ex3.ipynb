```python
import ee
```

```python
import geemap
```

```python
ee.Authenticate()
ee.Initialize(project="igneous-ethos-453102-d8")
```

```python
map=geemap.Map(ee_initialize=False)
map
```

```python
Taluk=ee.FeatureCollection('projects/igneous-ethos-453102-d8/assets/subdistrict_boundary')
map.addLayer(Taluk,{},'Taluk_Boundary')
map
```

```python
DEM=ee.ImageCollection("JAXA/ALOS/AW3D30/V4_1").select("DSM")
Taluk_DEM= DEM.filterBounds(Taluk).mosaic().clip(Taluk)
map.addLayer(Taluk_DEM, {"min":110,"max":207, "palette":["#FFF5E6", "#F7E0C3", "#EEC89A", "#E0B070", "#D29A52", "#C07F34", "#A8652A", "#8B4A20", "#6A3418", "#4A2410"]}, "DEM")
map
```

```python
task = ee.batch.Export.image.toDrive({
    'image': Taluk_DEM,
    'description': 'Taluk_DEM_Export',
    'folder': 'EarthEngine_Exports',
    'fileNamePrefix': 'Taluk_DEM',
    'scale': 30,
    'crs': 'EPSG:4326',
    'region': Taluk.geometry().bounds().getInfo()['coordinates'] # Exporting to the bounds of the Taluk feature collection
})

task.start()

print(f"Export task 'Taluk_DEM_Export' started. You can monitor its progress in the Earth Engine Tasks tab or your Google Drive folder 'EarthEngine_Exports'.")
```

```python
task = ee.batch.Export.image.toDrive(
    image=Taluk_DEM,
    description='Taluk_DEM_Export',
    folder='EarthEngine_Exports',
    fileNamePrefix='Taluk_DEM',
    scale=30,
    crs='EPSG:4326',
    region=Taluk.geometry().bounds().getInfo()['coordinates'] # Exporting to the bounds of the Taluk feature collection
)

task.start()

print(f"Export task 'Taluk_DEM_Export' started. You can monitor its progress in the Earth Engine Tasks tab or your Google Drive folder 'EarthEngine_Exports'.")
```

```python
NDVI = ee.ImageCollection("LANDSAT/COMPOSITES/C02/T1_L2_ANNUAL_NDVI")
TALUK_NDVI = NDVI.filterDate("2025-01-01","2025-12-31").mosaic().clip(Taluk)
map.addLayer(TALUK_NDVI,{"min":-0.3,"max":0.9,"palette":['#000080','#FFFF00','#FFAA00','#88FF00','#00FF00','#00AA00','#006600']},'TALUK_NDVI')
map
```

```python
task = ee.batch.Export.image.toDrive(
    image=TALUK_NDVI,
    description='TALUK_NDVI_Export',
    folder='EarthEngine_Exports',
    fileNamePrefix='TALUK_NDVI',
    scale=30,
    crs='EPSG:4326',
    region=Taluk.geometry().bounds().getInfo()['coordinates'] # Exporting to the bounds of the Taluk feature collection
)

task.start()

print(f"Export task 'TALUK_NDVI_Export' started. You can monitor its progress in the Earth Engine Tasks tab or your Google Drive folder 'EarthEngine_Exports'.")
```
